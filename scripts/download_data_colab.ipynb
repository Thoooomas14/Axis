# Axis Dataset Downloader (Partial/Small Batch)

This notebook downloads a **small subset** of the RT-X data (metadata + first few shards). This prevents running out of storage on Colab while still providing enough data to verify the training pipeline.

**Instructions:**
1. Run **Cell 1** to Authenticate.
2. Run **Cell 2** to download the subset.
3. Run **Cell 3** to zip.

```python
# 1. Authenticate User
from google.colab import auth
print("Authenticating...")
auth.authenticate_user()
print("Authentication successful!")
```

```python
# 2. Download Partial Data via CLI
import os

# Configuration
BUCKET_URI = 'gs://gresearch/robotics/fractal20220817_data/0.1.0'
LOCAL_DIR = '/content/axis_data/fractal20220817_data/0.1.0'

print(f"Downloading subset from {BUCKET_URI} to {LOCAL_DIR}...")
os.makedirs(LOCAL_DIR, exist_ok=True)

# 1. Download Metadata (Critical for TFDS to recognize the dataset)
print("Downloading metadata...")
!gcloud storage cp {BUCKET_URI}/dataset_info.json {LOCAL_DIR}/
!gcloud storage cp {BUCKET_URI}/features.json {LOCAL_DIR}/ 2>/dev/null || echo "No features.json found (optional)"

# 2. Download First Few Shards (e.g., first 5 files)
# The pattern *0000[0-4]-of-* matches 00000, 00001, 00002, 00003, 00004
print("Downloading data shards (Subset)...")
!gcloud storage cp "{BUCKET_URI}/*0000[0-4]-of-*" {LOCAL_DIR}/

print("Partial download complete!")
!ls -lh {LOCAL_DIR}
```

```python
# 3. Zip and Save
print("Zipping dataset...")
!zip -r /content/axis_data.zip /content/axis_data
print("Zipping complete. Download /content/axis_data.zip")
```
